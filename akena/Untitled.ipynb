```python
# If cv2 is not installed in the active notebook environment, run:
# !pip install opencv-python
import cv2

photo = cv2.imread("image2.jpg")
cv2.namedWindow("my window", cv2.WINDOW_NORMAL)  # create the window first
cv2.moveWindow("my window", 0, 0)  # then move it


cv2.imshow("my window", photo)
cv2.waitKey(5000)
cv2.destroyAllWindows()

```
