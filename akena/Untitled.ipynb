```python
import cv2
from ultralytics import YOLO

model = YOLO("yolov8n.pt")

photo = cv2.imread("image2.jpg")
if photo is None:
    raise FileNotFoundError("Could not load image2.jpg — check the path")

result = model(photo, show=True)  # YOLO opens its own window with boxes drawn

cv2.waitKey(50000)
cv2.destroyAllWindows()
```
